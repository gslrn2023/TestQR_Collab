```python
# 1. Download/Clone the GitHub project inside Google Colab
!git clone https://github.com/lincolnloop/python-qrcode.git

# 2. Change directory into the downloaded project folder
%cd python-qrcode

# 3. Install the project dependencies locally in Colab
!pip install -e .[pil]

# 4. Import the library from the cloned GitHub project and generate output
import qrcode

# Create a customized QR code
qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4,
)
qr.add_data("https://github.com/gslrn2023")
qr.make(fit=True)

# Generate and display the image directly inside Colab
# Change back_color to light yellow/orange
img = qr.make_image(fill_color="#1e3a8a", back_color="#fffbeb")
img
```
